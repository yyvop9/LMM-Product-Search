```python
from transformers import CLIPProcessor, CLIPModel
from PIL import Image
```

```python
# ----------------------------------------------------
# 1. 모델과 프로세서 로드
# (!!!) safetensors=True 옵션을 추가합니다!
# ----------------------------------------------------
print("CLIP 모델 로드를 시작합니다... (Safetensors 사용)")
try:
    model = CLIPModel.from_pretrained(
        "openai/clip-vit-base-patch32", 
        use_safetensors=True  # 👈 이 옵션 추가!
    )
    processor = CLIPProcessor.from_pretrained(
        "openai/clip-vit-base-patch32", 
        use_safetensors=True  # 👈 이 옵션 추가!
    )
    print("✅ CLIP 모델 로드 성공! (Safetensors 사용)")
except Exception as e:
    print(f"❌ 모델 로드 실패: {e}")

# ----------------------------------------------------
# 2. 테스트할 이미지와 텍스트 준비
# ----------------------------------------------------
# (경로 수정하신 것 그대로 사용)
image_path = "C:/LMM-Product-Search/data/images/1620.jpg" # (슬래시 '/'로 통일)
text_queries = ["파란색 셔츠 사진", "빨간색 바지 사진"]

try:
    # 3. 이미지 열기
    image = Image.open(image_path)
    print(f"✅ 이미지 '{image_path}' 열기 성공!")

    # 4. 이미지/텍스트를 벡터로 변환 (모델 입력 준비)
    inputs = processor(text=text_queries, images=image, return_tensors="pt", padding=True)

    # 5. 모델 실행 및 벡터 추출!
    outputs = model(**inputs)

    image_vector = outputs.image_embeds
    text_vectors = outputs.text_embeds

    print("\n--- 결과 ---")
    print(f"이미지 벡터 Shape: {image_vector.shape}")
    print(f"텍스트 벡터 Shape: {text_vectors.shape}")
    print(f"이미지 벡터 (앞 5개 값): {image_vector[0, :5].detach().numpy()}")
    print("🎉 성공! 이미지와 텍스트를 벡터로 변환했습니다.")

except FileNotFoundError:
    print(f"❌ 에러: '{image_path}' 파일을 찾을 수 없습니다.")
except Exception as e:
    print(f"❌ 벡터 변환 중 알 수 없는 에러: {e}")
```
